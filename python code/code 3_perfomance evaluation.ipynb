```python
# =========================================
# Notebook 03 — 2-Phase Evaluation (Full Corpus)
# =========================================
#
# CRITICAL FIX: Search corpus = FULL year corpus (14万~24万 patents),
#               NOT the 1050 sampled patents.
#
# Design:
#   Phase 1: Build rank caches
#     - Dense: embed ALL docs in full corpus → FAISS index → search with queries → top-L ranks
#     - BM25:  index ALL docs in full corpus → search with queries → top-L ranks
#     - Ranks are stored as integer indices into the full corpus doc list.
#     - STRICT cache validation via SHA1 hashes of corpus & query keys.
#     - Dense doc embedding supports RESUME via memmap + progress.json.
#   Phase 2: Evaluate all 12 models using cached ranks (fast)
#     - BM25-only, Dense-only, Hybrid (dense+BM25 weighted-RRF)
#     - Main weight 0.5:0.5; sensitivity 0.3:0.7 and 0.7:0.3
#     - Bootstrap CI (query-level, B=1000)
#
# NOTE on ground truth:
#   Each query's gold patent_id comes from queries_final.csv.
#   We locate this patent_id within the FULL corpus to get its doc index.
#   Queries whose gold patent_id is NOT found in the full corpus are flagged and excluded.

import os, re, json, time, math, pickle, random, hashlib
from dataclasses import dataclass
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import faiss

from sentence_transformers import SentenceTransformer

try:
    from FlagEmbedding import BGEM3FlagModel
    _HAS_BGE = True
except Exception as e:
    _HAS_BGE = False
    print("[WARN] FlagEmbedding import failed. bge-m3 dense retrieval will not run.")

try:
    from rank_bm25 import BM25Okapi
    _HAS_BM25 = True
except Exception as e:
    _HAS_BM25 = False
    print("[WARN] rank_bm25 import failed. BM25 retrieval will not run.")

from datetime import datetime

# -----------------------------------------
# 0) Global settings
# -----------------------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("DEVICE:", DEVICE)

YEARS = [2005, 2015, 2025]
QUERY_TYPES = ["summary", "keyword", "function"]

TOP_L = 100    # candidates per retriever
K_RRF = 60     # RRF constant

EVAL_K_LIST = [10, 50]
BOOTSTRAP_B = 1000

MAIN_WEIGHT = (0.5, 0.5)    # dense : BM25
SENS_WEIGHTS = [(0.3, 0.7), (0.7, 0.3)]

# -----------------------------------------
# 1) Directory policy
# -----------------------------------------
BASE_DIR = "2nd_exp"
DATA_DIR = os.path.join(BASE_DIR, "data")

CACHE_ROOT = os.path.join(BASE_DIR, "cache", "notebook03_rankcache_fullcorpus")
RESULTS_DIR = os.path.join(BASE_DIR, "results", "notebook03_eval_v2")
os.makedirs(CACHE_ROOT, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)

LOG_PATH = os.path.join(RESULTS_DIR, f"runlog_{datetime.now().strftime('%Y%m%d_%H%M%S')}.txt")

def log(msg: str):
    print(msg)
    with open(LOG_PATH, "a", encoding="utf-8") as f:
        f.write(msg + "\n")

log(f"DEVICE={DEVICE}")
log(f"CACHE_ROOT={CACHE_ROOT}")
log(f"RESULTS_DIR={RESULTS_DIR}")

# -----------------------------------------
# 2) Raw full-corpus CSV paths
#    (same files used by Notebook 01)
# -----------------------------------------
RAW_INPUT_FILES = {
    2005: "independent_claims_y2005_with_abstract_wipo_cpc.csv",
    2015: "independent_claims_y2015_with_abstract_wipo_cpc.csv",
    2025: "independent_claims_y2025_with_abstract_wipo_cpc.csv",
}

# -----------------------------------------
# 3) Utility functions
# -----------------------------------------
def normalize_text(s) -> str:
    if s is None:
        return ""
    s = str(s)
    s = s.replace("\r", " ").replace("\n", " ")
    s = re.sub(r"\s+", " ", s).strip()
    return s

def find_first_existing_col(df, candidates):
    cols_lower = {c.lower(): c for c in df.columns}
    for cand in candidates:
        if cand.lower() in cols_lower:
            return cols_lower[cand.lower()]
    return None

def sha1_of_list(str_list: List[str]) -> str:
    h = hashlib.sha1()
    for s in str_list:
        h.update(str(s).encode("utf-8"))
        h.update(b"\n")
    return h.hexdigest()

def simple_tokenize(text: str) -> List[str]:
    text = str(text).lower()
    return re.findall(r"[a-z0-9]+", text)

def safe_name(x: str) -> str:
    return re.sub(r"[^a-zA-Z0-9_\-]+", "_", str(x))

# -----------------------------------------
# 4) FULL CORPUS loading (KEY CHANGE)
# -----------------------------------------
@dataclass
class FullCorpus:
    year: int
    patent_ids: List[str]       # full corpus patent IDs
    claims: List[str]           # full corpus claims text
    abstract: List[str]         # full corpus abstract text
    n_docs: int
    corpus_sha1: str            # sha1 of patent_ids
    pid_to_idx: Dict[str, int]  # patent_id -> index in corpus

def load_full_corpus(year: int) -> FullCorpus:
    """Load the ENTIRE year corpus (14万~24万 patents), NOT just sampled 1050."""
    fname = RAW_INPUT_FILES[year]
    path = os.path.join(os.getcwd(), fname)
    if not os.path.exists(path):
        raise FileNotFoundError(f"[{year}] Raw corpus file not found: {fname}")

    log(f"[{year}] Loading FULL corpus: {fname}")
    df = pd.read_csv(path, dtype=str, keep_default_na=False, na_values=[], low_memory=False)
    log(f"[{year}] Full corpus loaded: rows={len(df):,}")

    # Map columns
    claims_col = find_first_existing_col(df, [
        "claims", "independent_claims", "independent_claim", "claim", "claims_text"
    ])
    abstract_col = find_first_existing_col(df, ["abstract", "abstract_text", "abs"])
    id_col = find_first_existing_col(df, [
        "patent_id", "publication_number", "pub_number", "doc_number", "patent_number", "id"
    ])

    if claims_col is None:
        raise ValueError(f"[{year}] Cannot find claims column. Cols: {list(df.columns)}")
    if abstract_col is None:
        raise ValueError(f"[{year}] Cannot find abstract column. Cols: {list(df.columns)}")
    if id_col is None:
        raise ValueError(f"[{year}] Cannot find patent_id column. Cols: {list(df.columns)}")

    if claims_col != "claims":
        df = df.rename(columns={claims_col: "claims"})
    if abstract_col != "abstract":
        df = df.rename(columns={abstract_col: "abstract"})
    if id_col != "patent_id":
        df = df.rename(columns={id_col: "patent_id"})

    # Normalize
    df["patent_id"] = df["patent_id"].astype(str).str.strip().str.replace(r"\.0$", "", regex=True)
    df["claims"] = df["claims"].fillna("").map(normalize_text)
    df["abstract"] = df["abstract"].fillna("").map(normalize_text)

    # Remove duplicates by patent_id (keep first)
    n_before = len(df)
    df = df.drop_duplicates(subset="patent_id", keep="first").reset_index(drop=True)
    if len(df) < n_before:
        log(f"[{year}] Removed {n_before - len(df)} duplicate patent_ids")

    pids = df["patent_id"].tolist()
    claims = df["claims"].tolist()
    abstract = df["abstract"].tolist()
    pid_to_idx = {pid: i for i, pid in enumerate(pids)}

    csha = sha1_of_list(pids)
    log(f"[{year}] Full corpus ready: n_docs={len(pids):,}, sha1={csha[:12]}")

    return FullCorpus(
        year=year,
        patent_ids=pids,
        claims=claims,
        abstract=abstract,
        n_docs=len(pids),
        corpus_sha1=csha,
        pid_to_idx=pid_to_idx,
    )

# -----------------------------------------
# 5) Query loading (from Notebook 02 output)
# -----------------------------------------
def load_queries(year: int) -> pd.DataFrame:
    candidates = [
        os.path.join(DATA_DIR, str(year), "queries_final.csv"),
        os.path.join(DATA_DIR, str(year), "queries.csv"),
    ]
    qpath = None
    for p in candidates:
        if os.path.exists(p):
            qpath = p
            break
    if qpath is None:
        raise FileNotFoundError(f"[{year}] queries file not found. Tried: {candidates}")

    dfq = pd.read_csv(qpath, dtype={"patent_id": str})
    dfq["patent_id"] = dfq["patent_id"].astype(str).str.strip().str.replace(r"\.0$", "", regex=True)
    dfq["query_type"] = dfq["query_type"].astype(str).str.lower().str.strip()
    dfq["query_text"] = dfq["query_text"].fillna("").astype(str)

    dfq = dfq[dfq["query_type"].isin(QUERY_TYPES)].reset_index(drop=True)

    if "query_key" not in dfq.columns:
        dfq["query_key"] = [f"{year}::{i}" for i in range(len(dfq))]

    dfq = dfq.sort_values(["query_type", "patent_id", "query_key"]).reset_index(drop=True)

    return dfq

def validate_queries_against_corpus(dfq: pd.DataFrame, corpus: FullCorpus, year: int) -> pd.DataFrame:
    """Check that each query's gold patent_id exists in the full corpus.
    Remove queries whose gold patent is missing."""
    mask = dfq["patent_id"].isin(corpus.pid_to_idx)
    n_bad = (~mask).sum()
    if n_bad > 0:
        log(f"[{year}] WARNING: {n_bad} queries have gold patent_id not found in full corpus. Removing them.")
        bad_pids = dfq.loc[~mask, "patent_id"].unique().tolist()
        log(f"  Missing PIDs (first 10): {bad_pids[:10]}")
    dfq = dfq[mask].reset_index(drop=True)
    return dfq

# -----------------------------------------
# 6) Model configs (12 models)
# -----------------------------------------
MODEL_CONFIGS = [
    {"model_id": "BGE_M3_512_CLAIMS_DENSE",  "display_name": "BGE-M3_512_Claims_Dense",
     "field": "claims", "retrieval": "dense", "encoder": "BAAI/bge-m3", "max_length": 512},
    {"model_id": "BGE_M3_512_CLAIMS_HYBRID", "display_name": "BGE-M3_512_Claims_Hybrid",
     "field": "claims", "retrieval": "hybrid", "encoder": "BAAI/bge-m3", "max_length": 512},

    {"model_id": "BGE_M3_512_ABSTRACT_DENSE",  "display_name": "BGE-M3_512_Abstract_Dense",
     "field": "abstract", "retrieval": "dense", "encoder": "BAAI/bge-m3", "max_length": 512},
    {"model_id": "BGE_M3_512_ABSTRACT_HYBRID", "display_name": "BGE-M3_512_Abstract_Hybrid",
     "field": "abstract", "retrieval": "hybrid", "encoder": "BAAI/bge-m3", "max_length": 512},

    {"model_id": "PATENTSBERTA_CLAIMS_DENSE",  "display_name": "PatentSBERTa_Claims_Dense",
     "field": "claims", "retrieval": "dense", "encoder": "AI-Growth-Lab/PatentSBERTa", "max_length": 512},
    {"model_id": "PATENTSBERTA_CLAIMS_HYBRID", "display_name": "PatentSBERTa_Claims_Hybrid",
     "field": "claims", "retrieval": "hybrid", "encoder": "AI-Growth-Lab/PatentSBERTa", "max_length": 512},

    {"model_id": "PATENTSBERTA_ABSTRACT_DENSE",  "display_name": "PatentSBERTa_Abstract_Dense",
     "field": "abstract", "retrieval": "dense", "encoder": "AI-Growth-Lab/PatentSBERTa", "max_length": 512},
    {"model_id": "PATENTSBERTA_ABSTRACT_HYBRID", "display_name": "PatentSBERTa_Abstract_Hybrid",
     "field": "abstract", "retrieval": "hybrid", "encoder": "AI-Growth-Lab/PatentSBERTa", "max_length": 512},

    {"model_id": "BM25_CLAIMS",   "display_name": "BM25_Claims",
     "field": "claims", "retrieval": "bm25", "encoder": None, "max_length": None},
    {"model_id": "BM25_ABSTRACT", "display_name": "BM25_Abstract",
     "field": "abstract", "retrieval": "bm25", "encoder": None, "max_length": None},
]
assert len(MODEL_CONFIGS) == 10
log(f"Total models: {len(MODEL_CONFIGS)}")

# -----------------------------------------
# 7) Rank cache I/O
# -----------------------------------------
def rankcache_paths(year: int, kind: str, field: str, encoder: Optional[str] = None,
                    max_length: Optional[int] = None) -> Tuple[str, str]:
    ydir = os.path.join(CACHE_ROOT, str(year), "rankcache")
    os.makedirs(ydir, exist_ok=True)
    if kind == "bm25":
        base = f"{kind}__{field}__L{TOP_L}"
    else:
        enc = safe_name(encoder)
        base = f"{kind}__{field}__{enc}__ML{int(max_length)}__L{TOP_L}"
    return (os.path.join(ydir, base + ".meta.json"),
            os.path.join(ydir, base + ".ranks.npy"))

def save_rankcache(meta_path, ranks_path, meta, ranks):
    tmp = meta_path + ".tmp"
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)
    for attempt in range(5):
        try:
            os.replace(tmp, meta_path)
            break
        except PermissionError:
            time.sleep(0.5)
    else:
        with open(meta_path, "w", encoding="utf-8") as f:
            json.dump(meta, f, ensure_ascii=False, indent=2)
        if os.path.exists(tmp):
            try:
                os.remove(tmp)
            except Exception:
                pass
    np.save(ranks_path, ranks.astype(np.int32))

def load_rankcache(meta_path, ranks_path):
    with open(meta_path, "r", encoding="utf-8") as f:
        meta = json.load(f)
    ranks = np.load(ranks_path, allow_pickle=False)
    return meta, ranks

def validate_rankcache_meta(meta, expected):
    for k, v in expected.items():
        if meta.get(k) != v:
            return (False, f"meta mismatch: {k} (cache={meta.get(k)} vs expected={v})")
    return (True, "ok")

# -----------------------------------------
# 8) DenseIndex with resume (memmap + progress)
# -----------------------------------------
class DenseIndex:
    def __init__(self, year, encoder, max_length, field, corpus_texts):
        self.year = year
        self.encoder = encoder
        self.max_length = int(max_length)
        self.field = field
        self.corpus_texts = corpus_texts

        self.year_dir = os.path.join(CACHE_ROOT, str(year), "dense_index")
        os.makedirs(self.year_dir, exist_ok=True)

        enc_safe = safe_name(encoder)
        self.emb_mmap_path = os.path.join(self.year_dir, f"docemb__{enc_safe}__{field}__ML{self.max_length}.mmap")
        self.emb_progress_path = os.path.join(self.year_dir, f"docemb__{enc_safe}__{field}__ML{self.max_length}.progress.json")
        self.index_path = os.path.join(self.year_dir, f"faiss__{enc_safe}__{field}__ML{self.max_length}.index")

        self.index = None
        self.doc_emb = None
        self.dim = None

        if "bge-m3" in encoder.lower():
            if not _HAS_BGE:
                raise RuntimeError("FlagEmbedding not available.")
            self.is_bge = True
            self.model = BGEM3FlagModel("BAAI/bge-m3", use_fp16=True, device=DEVICE)
        else:
            self.is_bge = False
            self.model = SentenceTransformer(encoder, device=DEVICE)
            self.model.max_seq_length = self.max_length

    def _encode_texts(self, texts, batch_size):
        if self.is_bge:
            out = self.model.encode(texts, batch_size=batch_size, max_length=self.max_length)
            if isinstance(out, dict):
                out = out.get("dense_vecs", list(out.values())[0])
            emb = np.asarray(out, dtype=np.float32)
        else:
            emb = self.model.encode(
                texts, batch_size=batch_size, show_progress_bar=False, convert_to_numpy=True
            ).astype(np.float32)
        faiss.normalize_L2(emb)
        return emb

    def _load_progress(self):
        if os.path.exists(self.emb_progress_path):
            with open(self.emb_progress_path, "r", encoding="utf-8") as f:
                return json.load(f)
        return {}

    def _save_progress(self, d):
        tmp = self.emb_progress_path + ".tmp"
        with open(tmp, "w", encoding="utf-8") as f:
            json.dump(d, f, ensure_ascii=False, indent=2)
        for attempt in range(5):
            try:
                os.replace(tmp, self.emb_progress_path)
                return
            except PermissionError:
                time.sleep(0.5)
        # fallback: 직접 덮어쓰기
        with open(self.emb_progress_path, "w", encoding="utf-8") as f:
            json.dump(d, f, ensure_ascii=False, indent=2)
        if os.path.exists(tmp):
            try:
                os.remove(tmp)
            except Exception:
                pass

    def build_or_load(self, doc_batch_size, chunk_docs=256):
        n_docs = len(self.corpus_texts)
        prog = self._load_progress()

        if os.path.exists(self.emb_mmap_path) and prog.get("status") == "done" and prog.get("n_docs") == n_docs:
            self.dim = int(prog["dim"])
            self.doc_emb = np.memmap(self.emb_mmap_path, dtype="float32", mode="r", shape=(n_docs, self.dim))
            log(f"  [DenseIndex] docemb ready (cached): year={self.year} field={self.field} enc={self.encoder} ML={self.max_length} shape=({n_docs},{self.dim})")
        else:
            next_idx = int(prog.get("next_idx", 0))
            dim = prog.get("dim", None)

            # Validate that existing progress matches current corpus size
            if prog.get("n_docs") is not None and prog.get("n_docs") != n_docs:
                log(f"  [DenseIndex] WARNING: n_docs mismatch (progress={prog.get('n_docs')} vs current={n_docs}). Rebuilding from scratch.")
                next_idx = 0
                dim = None

            log(f"  [DenseIndex] Building docemb: year={self.year} field={self.field} enc={self.encoder} ML={self.max_length}")
            log(f"              resume next_idx={next_idx}, dim={dim}, n_docs={n_docs}")

            if dim is None:
                first_n = min(chunk_docs, n_docs)
                first = self._encode_texts(self.corpus_texts[:first_n], batch_size=min(doc_batch_size, first_n))
                dim = int(first.shape[1])
                self.dim = dim
                self.doc_emb = np.memmap(self.emb_mmap_path, dtype="float32", mode="w+", shape=(n_docs, dim))
                self.doc_emb[:first.shape[0]] = first
                self.doc_emb.flush()
                next_idx = first.shape[0]
                self._save_progress({"status": "building", "dim": dim, "next_idx": next_idx, "n_docs": n_docs,
                    "encoder": self.encoder, "max_length": self.max_length, "field": self.field,
                    "updated_at": datetime.now().isoformat()})
                log(f"              initialized mmap + wrote first chunk [0:{next_idx}] dim={dim}")
            else:
                self.dim = int(dim)
                mode = "r+" if os.path.exists(self.emb_mmap_path) else "w+"
                self.doc_emb = np.memmap(self.emb_mmap_path, dtype="float32", mode=mode, shape=(n_docs, self.dim))

            for s in range(next_idx, n_docs, chunk_docs):
                e = min(s + chunk_docs, n_docs)
                emb = self._encode_texts(self.corpus_texts[s:e], batch_size=min(doc_batch_size, e - s))
                self.doc_emb[s:e] = emb
                self.doc_emb.flush()
                self._save_progress({"status": "building", "dim": self.dim, "next_idx": e, "n_docs": n_docs,
                    "encoder": self.encoder, "max_length": self.max_length, "field": self.field,
                    "updated_at": datetime.now().isoformat()})
                if (s // chunk_docs) % 20 == 0:
                    log(f"              docemb [{s}:{e}] / {n_docs}")

            self._save_progress({"status": "done", "dim": self.dim, "next_idx": n_docs, "n_docs": n_docs,
                "encoder": self.encoder, "max_length": self.max_length, "field": self.field,
                "updated_at": datetime.now().isoformat()})
            log(f"  [DenseIndex] docemb DONE: {self.emb_mmap_path}")

        # FAISS index
        if os.path.exists(self.index_path):
            self.index = faiss.read_index(self.index_path)
            if self.index.ntotal != n_docs:
                log(f"  [DenseIndex] FAISS ntotal mismatch ({self.index.ntotal} vs {n_docs}). Rebuilding.")
                self.index = None
            else:
                log(f"  [DenseIndex] faiss index loaded: {self.index_path} (ntotal={self.index.ntotal})")

        if self.index is None:
            log(f"  [DenseIndex] Building FAISS index for {n_docs} docs...")
            index = faiss.IndexFlatIP(self.dim)
            # Add in chunks to avoid memory spike
            add_chunk = 50000
            for s in range(0, n_docs, add_chunk):
                e = min(s + add_chunk, n_docs)
                index.add(np.asarray(self.doc_emb[s:e], dtype=np.float32))
            faiss.write_index(index, self.index_path)
            self.index = index
            log(f"  [DenseIndex] faiss index built: {self.index_path} (ntotal={self.index.ntotal})")

    def search_batch(self, query_texts, top_l, query_batch_size, log_every=500):
        all_idx = []
        n = len(query_texts)
        for s in range(0, n, query_batch_size):
            chunk = query_texts[s:s+query_batch_size]
            qemb = self._encode_texts(chunk, batch_size=min(query_batch_size, len(chunk)))
            _, idxs = self.index.search(qemb, top_l)
            all_idx.append(idxs.astype(np.int32))
            done = min(s + query_batch_size, n)
            if done % log_every < query_batch_size or done == n:
                log(f"    [Dense search] {done}/{n} queries done")
        return np.vstack(all_idx)

# -----------------------------------------
# 9) Phase 1 — Build rank caches
#    Key: corpus = FULL (14万~24万 docs)
#         queries = from Notebook 02 (~3150 per year)
# -----------------------------------------
def build_bm25_ranks(corpus: FullCorpus, dfq: pd.DataFrame, field: str) -> np.ndarray:
    if not _HAS_BM25:
        raise RuntimeError("rank_bm25 not available")

    texts = corpus.claims if field == "claims" else corpus.abstract
    log(f"  [BM25] Tokenizing {len(texts):,} docs for field={field}...")
    tokenized = [simple_tokenize(t) for t in texts]
    log(f"  [BM25] Building BM25 index...")
    bm25 = BM25Okapi(tokenized)

    qtexts = dfq["query_text"].tolist()
    n_q = len(qtexts)
    ranks = np.empty((n_q, TOP_L), dtype=np.int32)

    log(f"  [BM25] Searching: n_queries={n_q}, topL={TOP_L}")
    for i, qt in enumerate(qtexts):
        scores = bm25.get_scores(simple_tokenize(qt))
        top_idx = np.argsort(scores)[::-1][:TOP_L]
        ranks[i, :] = top_idx.astype(np.int32)
        if (i + 1) % 500 == 0 or i == n_q - 1:
            log(f"    [BM25] {i+1}/{n_q}")

    return ranks

def build_dense_ranks(corpus: FullCorpus, dfq: pd.DataFrame, field: str,
                      encoder: str, max_length: int,
                      doc_batch_size: int, query_batch_size: int) -> np.ndarray:
    texts = corpus.claims if field == "claims" else corpus.abstract

    dindex = DenseIndex(year=corpus.year, encoder=encoder, max_length=max_length,
                        field=field, corpus_texts=texts)

    chunk_docs = 256
    dindex.build_or_load(doc_batch_size=doc_batch_size, chunk_docs=chunk_docs)

    qtexts = dfq["query_text"].tolist()
    log(f"  [Dense] Searching: enc={encoder} ML={max_length} field={field} n_queries={len(qtexts)} topL={TOP_L}")
    ranks = dindex.search_batch(qtexts, top_l=TOP_L, query_batch_size=query_batch_size)
    return ranks.astype(np.int32)

def phase1_build_rankcaches(corpus: FullCorpus, dfq: pd.DataFrame):
    year = corpus.year
    query_keys = dfq["query_key"].tolist()
    query_keys_sha1 = sha1_of_list(query_keys)
    corpus_sha1 = corpus.corpus_sha1
    n_q = len(dfq)
    n_docs = corpus.n_docs

    log(f"\n=== Phase 1: year={year} ===")
    log(f"  FULL corpus: n_docs={n_docs:,}, sha1={corpus_sha1[:12]}")
    log(f"  queries: n_q={n_q}, sha1={query_keys_sha1[:12]}")

    # ---- BM25 ranks ----
    bm25_ranks = {}
    for field in ["claims", "abstract"]:
        meta_path, ranks_path = rankcache_paths(year, kind="bm25", field=field)
        expected = {
            "year": year, "kind": "bm25", "field": field, "top_l": TOP_L,
            "n_queries": n_q, "n_docs": n_docs,
            "query_keys_sha1": query_keys_sha1, "corpus_sha1": corpus_sha1,
        }
        if os.path.exists(meta_path) and os.path.exists(ranks_path):
            meta, ranks = load_rankcache(meta_path, ranks_path)
            ok, reason = validate_rankcache_meta(meta, expected)
            if ok:
                bm25_ranks[field] = ranks
                log(f"  [BM25] cache hit: field={field}, shape={ranks.shape}")
                continue
            else:
                log(f"  [BM25] cache invalid ({reason}) → rebuild")
        else:
            log(f"  [BM25] cache miss → build. field={field}")

        ranks = build_bm25_ranks(corpus, dfq, field)
        meta = {**expected, "built_at": datetime.now().isoformat()}
        save_rankcache(meta_path, ranks_path, meta, ranks)
        bm25_ranks[field] = ranks
        log(f"  [BM25] saved: {ranks_path}")

    # ---- Dense ranks ----
    needed = set()
    for cfg in MODEL_CONFIGS:
        if cfg["retrieval"] in ("dense", "hybrid"):
            needed.add((cfg["encoder"], int(cfg["max_length"]), cfg["field"]))

    dense_ranks = {}
    for (encoder, max_len, field) in sorted(needed):
        meta_path, ranks_path = rankcache_paths(year, kind="dense", field=field, encoder=encoder, max_length=max_len)
        expected = {
            "year": year, "kind": "dense", "field": field,
            "encoder": encoder, "max_length": int(max_len), "top_l": TOP_L,
            "n_queries": n_q, "n_docs": n_docs,
            "query_keys_sha1": query_keys_sha1, "corpus_sha1": corpus_sha1,
        }
        if os.path.exists(meta_path) and os.path.exists(ranks_path):
            meta, ranks = load_rankcache(meta_path, ranks_path)
            ok, reason = validate_rankcache_meta(meta, expected)
            if ok:
                dense_ranks[(encoder, max_len, field)] = ranks
                log(f"  [Dense] cache hit: field={field} enc={encoder} ML={max_len}, shape={ranks.shape}")
                continue
            else:
                log(f"  [Dense] cache invalid ({reason}) → rebuild")
        else:
            log(f"  [Dense] cache miss → build. field={field} enc={encoder} ML={max_len}")

        doc_bs = 32 if "bge-m3" in encoder.lower() else 64
        q_bs = 32

        ranks = build_dense_ranks(corpus, dfq, field, encoder, max_len, doc_bs, q_bs)
        meta = {**expected, "built_at": datetime.now().isoformat(), "doc_batch_size": doc_bs, "query_batch_size": q_bs}
        save_rankcache(meta_path, ranks_path, meta, ranks)
        dense_ranks[(encoder, max_len, field)] = ranks
        log(f"  [Dense] saved: {ranks_path}")

    return bm25_ranks, dense_ranks

# -----------------------------------------
# 10) Phase 2 — Evaluation (weighted-RRF)
# -----------------------------------------
def weighted_rrf_fuse(dense_rank, bm25_rank, w_dense, w_sparse, k_rrf, out_topk):
    scores = {}
    for r, doc in enumerate(dense_rank, start=1):
        scores[int(doc)] = scores.get(int(doc), 0.0) + (w_dense / (k_rrf + r))
    for r, doc in enumerate(bm25_rank, start=1):
        scores[int(doc)] = scores.get(int(doc), 0.0) + (w_sparse / (k_rrf + r))
    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    return np.array([doc for doc, _ in ranked[:out_topk]], dtype=np.int32)

def hit_at_k(ranked_pids, gold_pid, k):
    return 1.0 if gold_pid in ranked_pids[:k] else 0.0

def mrr_at_k(ranked_pids, gold_pid, k):
    for i, pid in enumerate(ranked_pids[:k], start=1):
        if pid == gold_pid:
            return 1.0 / i
    return 0.0

def compute_metrics(ranked_pids, gold_pid):
    return {
        "Hit@10": hit_at_k(ranked_pids, gold_pid, 10),
        "MRR@10": mrr_at_k(ranked_pids, gold_pid, 10),
        "Hit@50": hit_at_k(ranked_pids, gold_pid, 50),
    }

def bootstrap_ci(values, B=1000, seed=42):
    rng = np.random.default_rng(seed)
    n = len(values)
    if n == 0:
        return (np.nan, np.nan)
    samples = rng.choice(values, size=(B, n), replace=True)
    means = samples.mean(axis=1)
    return (float(np.percentile(means, 2.5)), float(np.percentile(means, 97.5)))

@dataclass
class YearCache:
    year: int
    corpus: FullCorpus
    dfq: pd.DataFrame
    bm25_ranks: Dict[str, np.ndarray]
    dense_ranks: Dict[Tuple, np.ndarray]

def evaluate_one_cell(yc: YearCache, query_type: str, cfg: dict, weight: Tuple[float, float]) -> dict:
    df_sub = yc.dfq[yc.dfq["query_type"] == query_type].reset_index(drop=True)
    # Get indices in the FULL dfq (to slice rank arrays)
    idxs = yc.dfq.index[yc.dfq["query_type"] == query_type].to_numpy()
    n = len(df_sub)

    if n == 0:
        return {
            "year": yc.year, "query_type": query_type,
            "model_id": cfg["model_id"], "display_name": cfg["display_name"],
            "n_queries": 0,
            "Hit@10": np.nan, "MRR@10": np.nan, "Hit@50": np.nan,
            "Hit@10_ci_lo": np.nan, "Hit@10_ci_hi": np.nan,
            "MRR@10_ci_lo": np.nan, "MRR@10_ci_hi": np.nan,
            "Hit@50_ci_lo": np.nan, "Hit@50_ci_hi": np.nan,
            "weight_dense": weight[0], "weight_sparse": weight[1],
        }

    gold_pids = yc.dfq.loc[idxs, "patent_id"].astype(str).tolist()
    docid_to_pid = yc.corpus.patent_ids  # full corpus

    per_hit10 = np.zeros(n)
    per_mrr10 = np.zeros(n)
    per_hit50 = np.zeros(n)

    retrieval = cfg["retrieval"]
    field = cfg["field"]

    if retrieval == "bm25":
        top_docs = yc.bm25_ranks[field][idxs, :50]
        for i in range(n):
            ranked_pids = [docid_to_pid[int(d)] for d in top_docs[i]]
            m = compute_metrics(ranked_pids, gold_pids[i])
            per_hit10[i] = m["Hit@10"]; per_mrr10[i] = m["MRR@10"]; per_hit50[i] = m["Hit@50"]

    elif retrieval == "dense":
        encoder = cfg["encoder"]; max_len = int(cfg["max_length"])
        top_docs = yc.dense_ranks[(encoder, max_len, field)][idxs, :50]
        for i in range(n):
            ranked_pids = [docid_to_pid[int(d)] for d in top_docs[i]]
            m = compute_metrics(ranked_pids, gold_pids[i])
            per_hit10[i] = m["Hit@10"]; per_mrr10[i] = m["MRR@10"]; per_hit50[i] = m["Hit@50"]

    elif retrieval == "hybrid":
        encoder = cfg["encoder"]; max_len = int(cfg["max_length"])
        dense_r = yc.dense_ranks[(encoder, max_len, field)][idxs, :TOP_L]
        bm25_r = yc.bm25_ranks[field][idxs, :TOP_L]
        wd, ws = weight
        for i in range(n):
            fused = weighted_rrf_fuse(dense_r[i], bm25_r[i], wd, ws, K_RRF, out_topk=50)
            ranked_pids = [docid_to_pid[int(d)] for d in fused]
            m = compute_metrics(ranked_pids, gold_pids[i])
            per_hit10[i] = m["Hit@10"]; per_mrr10[i] = m["MRR@10"]; per_hit50[i] = m["Hit@50"]
    else:
        raise ValueError(f"Unknown retrieval: {retrieval}")

    hit10_ci = bootstrap_ci(per_hit10, B=BOOTSTRAP_B, seed=SEED)
    mrr10_ci = bootstrap_ci(per_mrr10, B=BOOTSTRAP_B, seed=SEED)
    hit50_ci = bootstrap_ci(per_hit50, B=BOOTSTRAP_B, seed=SEED)

    return {
        "year": yc.year, "query_type": query_type,
        "model_id": cfg["model_id"], "display_name": cfg["display_name"],
        "n_queries": n,
        "Hit@10": float(per_hit10.mean()), "MRR@10": float(per_mrr10.mean()), "Hit@50": float(per_hit50.mean()),
        "Hit@10_ci_lo": hit10_ci[0], "Hit@10_ci_hi": hit10_ci[1],
        "MRR@10_ci_lo": mrr10_ci[0], "MRR@10_ci_hi": mrr10_ci[1],
        "Hit@50_ci_lo": hit50_ci[0], "Hit@50_ci_hi": hit50_ci[1],
        "weight_dense": weight[0], "weight_sparse": weight[1],
    }

def macro_avg(df, col):
    x = df[col].dropna().values
    return float(np.mean(x)) if len(x) else np.nan

# =========================================
# 11) MAIN EXECUTION
# =========================================
t0 = time.time()

# --- Load full corpora + queries ---
CORPUS_BY_YEAR = {}
QUERIES_BY_YEAR = {}

for y in YEARS:
    log(f"\n--- Loading year={y} ---")
    CORPUS_BY_YEAR[y] = load_full_corpus(y)
    dfq = load_queries(y)
    dfq = validate_queries_against_corpus(dfq, CORPUS_BY_YEAR[y], y)
    QUERIES_BY_YEAR[y] = dfq
    log(f"  FULL corpus: {CORPUS_BY_YEAR[y].n_docs:,} docs")
    log(f"  queries: {len(dfq)} (types={sorted(dfq['query_type'].unique().tolist())})")

# --- Phase 1: Build rank caches ---
PHASE1 = {}
for y in YEARS:
    bm25_r, dense_r = phase1_build_rankcaches(CORPUS_BY_YEAR[y], QUERIES_BY_YEAR[y])
    PHASE1[y] = (bm25_r, dense_r)

phase1_elapsed = time.time() - t0
log(f"\nPhase 1 done. elapsed={phase1_elapsed/60:.2f} min")

# --- Phase 2: Evaluate ---
t1 = time.time()

YEARCACHE = {}
for y in YEARS:
    bm25_r, dense_r = PHASE1[y]
    YEARCACHE[y] = YearCache(
        year=y, corpus=CORPUS_BY_YEAR[y], dfq=QUERIES_BY_YEAR[y],
        bm25_ranks=bm25_r, dense_ranks=dense_r
    )

# Main weight evaluation
CELL_RESULTS = []
log("\n=== Evaluating (MAIN weight 0.5/0.5) ===")
for y in YEARS:
    yc = YEARCACHE[y]
    for qt in QUERY_TYPES:
        for cfg in MODEL_CONFIGS:
            res = evaluate_one_cell(yc, qt, cfg, weight=MAIN_WEIGHT)
            CELL_RESULTS.append(res)
        df_tmp = pd.DataFrame([r for r in CELL_RESULTS if r["year"]==y and r["query_type"]==qt])
        if len(df_tmp):
            best = df_tmp.sort_values("Hit@10", ascending=False).iloc[0]
            log(f"  [{y}][{qt}] best: {best['display_name']} Hit@10={best['Hit@10']:.4f}")

df_cell = pd.DataFrame(CELL_RESULTS)
cell_path = os.path.join(RESULTS_DIR, "results_cellwise_mainweight.csv")
df_cell.to_csv(cell_path, index=False)
log(f"Saved: {cell_path}")

phase2_elapsed = time.time() - t1
log(f"\nPhase 2 (main eval) done. elapsed={phase2_elapsed/60:.2f} min")

# -----------------------------------------
# 12) Reporting: Overall pooled + Worst-year
# -----------------------------------------
pooled_rows = []
for cfg in MODEL_CONFIGS:
    sub = df_cell[df_cell["model_id"] == cfg["model_id"]]
    pooled_rows.append({
        "model_id": cfg["model_id"], "display_name": cfg["display_name"],
        "Overall_Hit@10": macro_avg(sub, "Hit@10"),
        "Overall_MRR@10": macro_avg(sub, "MRR@10"),
        "Overall_Hit@50": macro_avg(sub, "Hit@50"),
    })
df_overall = pd.DataFrame(pooled_rows)

worst_rows = []
for cfg in MODEL_CONFIGS:
    per_year = []
    for y in YEARS:
        suby = df_cell[(df_cell["model_id"] == cfg["model_id"]) & (df_cell["year"] == y)]
        per_year.append({
            "year": y,
            "Hit@10": macro_avg(suby, "Hit@10"),
            "MRR@10": macro_avg(suby, "MRR@10"),
            "Hit@50": macro_avg(suby, "Hit@50"),
        })
    df_py = pd.DataFrame(per_year)
    worst_rows.append({
        "model_id": cfg["model_id"], "display_name": cfg["display_name"],
        "WorstYear_Hit@10": float(df_py["Hit@10"].min()),
        "WorstYear_MRR@10": float(df_py["MRR@10"].min()),
        "WorstYear_Hit@50": float(df_py["Hit@50"].min()),
    })
df_worst = pd.DataFrame(worst_rows)

df_summary = df_overall.merge(df_worst, on=["model_id","display_name"], how="left")
df_summary = df_summary.sort_values("Overall_Hit@10", ascending=False).reset_index(drop=True)

summary_path = os.path.join(RESULTS_DIR, "table_main_optionA_overall_and_worst.csv")
df_summary.to_csv(summary_path, index=False)
log(f"Saved: {summary_path}")

manu_path = os.path.join(RESULTS_DIR, "manuscript_table_optionA.csv")
main_cols = ["display_name","Overall_Hit@10","Overall_MRR@10","Overall_Hit@50",
             "WorstYear_Hit@10","WorstYear_MRR@10","WorstYear_Hit@50"]
df_summary[main_cols].to_csv(manu_path, index=False)
log(f"Saved: {manu_path}")

log("\n--- Manuscript Table preview ---")
log(df_summary[main_cols].to_markdown(index=False))

# -----------------------------------------
# 13) Appendix: year-wise table
# -----------------------------------------
rows_year = []
for cfg in MODEL_CONFIGS:
    for y in YEARS:
        sub = df_cell[(df_cell["model_id"] == cfg["model_id"]) & (df_cell["year"] == y)]
        rows_year.append({
            "year": y, "model_id": cfg["model_id"], "display_name": cfg["display_name"],
            "Hit@10": macro_avg(sub, "Hit@10"),
            "MRR@10": macro_avg(sub, "MRR@10"),
            "Hit@50": macro_avg(sub, "Hit@50"),
        })
df_by_year = pd.DataFrame(rows_year).sort_values(["year","Hit@10"], ascending=[True, False])
by_year_path = os.path.join(RESULTS_DIR, "appendix_table_by_year_macro_over_types.csv")
df_by_year.to_csv(by_year_path, index=False)
log(f"Saved: {by_year_path}")

# Plot
topN = 6
top_models = df_summary["model_id"].head(topN).tolist()
if "BM25_CLAIMS" not in top_models:
    top_models.append("BM25_CLAIMS")

plot_df = df_by_year[df_by_year["model_id"].isin(top_models)].copy()
plt.figure(figsize=(8,4))
for mid in top_models:
    sub = plot_df[plot_df["model_id"] == mid].sort_values("year")
    if len(sub):
        plt.plot(sub["year"].values, sub["Hit@10"].values, marker="o", label=sub["display_name"].iloc[0])
plt.xticks(YEARS)
plt.ylim(0, 1)
plt.xlabel("Year")
plt.ylabel("Hit@10 (macro over query types)")
plt.title("Hit@10 by Year (Selected Models)")
plt.legend(fontsize=8, loc="best")
fig_path = os.path.join(RESULTS_DIR, "appendix_fig_hit10_by_year_selected.png")
plt.tight_layout()
plt.savefig(fig_path, dpi=300)
plt.show()
log(f"Saved: {fig_path}")

# -----------------------------------------
# 14) Sensitivity analysis (hybrid weight variants)
# -----------------------------------------
HYBRID_MODEL_IDS = [cfg["model_id"] for cfg in MODEL_CONFIGS if cfg["retrieval"] == "hybrid"]

SENS_CELL_RESULTS = []
log("\n=== Sensitivity analysis (hybrid weights) ===")
for w in SENS_WEIGHTS:
    wd, ws = w
    log(f"\n--- Weight={wd:.1f}/{ws:.1f} ---")
    for y in YEARS:
        yc = YEARCACHE[y]
        for qt in QUERY_TYPES:
            for cfg in MODEL_CONFIGS:
                if cfg["model_id"] not in HYBRID_MODEL_IDS:
                    continue
                res = evaluate_one_cell(yc, qt, cfg, weight=w)
                SENS_CELL_RESULTS.append(res)

df_sens = pd.DataFrame(SENS_CELL_RESULTS)
sens_path = os.path.join(RESULTS_DIR, "appendix_sensitivity_cellwise_hybridweights.csv")
df_sens.to_csv(sens_path, index=False)
log(f"Saved: {sens_path}")

pooled_sens = []
for (wd, ws) in SENS_WEIGHTS:
    subw = df_sens[(df_sens["weight_dense"] == wd) & (df_sens["weight_sparse"] == ws)]
    for mid in HYBRID_MODEL_IDS:
        subm = subw[subw["model_id"] == mid]
        pooled_sens.append({
            "weight_dense": wd, "weight_sparse": ws,
            "model_id": mid,
            "display_name": subm["display_name"].iloc[0] if len(subm) else mid,
            "Overall_Hit@10": macro_avg(subm, "Hit@10"),
            "Overall_MRR@10": macro_avg(subm, "MRR@10"),
            "Overall_Hit@50": macro_avg(subm, "Hit@50"),
        })
df_pooled_sens = pd.DataFrame(pooled_sens).sort_values(["weight_dense","Overall_Hit@10"], ascending=[True, False])
sens_summary_path = os.path.join(RESULTS_DIR, "appendix_sensitivity_pooled_summary.csv")
df_pooled_sens.to_csv(sens_summary_path, index=False)
log(f"Saved: {sens_summary_path}")

log("\n--- Sensitivity pooled summary ---")
log(df_pooled_sens.to_markdown(index=False))

# -----------------------------------------
# 15) Final
# -----------------------------------------
total_elapsed = time.time() - t0
log(f"\nNotebook 03 finished. total={total_elapsed/60:.2f} min")
log("All done.")
```
